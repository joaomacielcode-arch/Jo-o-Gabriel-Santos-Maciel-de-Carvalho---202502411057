```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("livros.csv", sep =";")
```

```python
# Susbtitui todos os '0' do df por NaN's
df.replace(0,np.nan,inplace=True)
# Fiz essa substituição porque o fillna() não troca os valores com 0, só NaN
```



---
*   Modificando o dataframe:





```python
# Susbstituindo os anos nulos pela mediana (2012)
df["ano"] = df["ano"].fillna(df["ano"].median).astype("int32")
```

```python
# Susbstituindo os autores nulos por "Sem Crédito"
df["autor"] = df["autor"].fillna("Sem Crédito").astype("object")
```

```python
df["faixa_paginas"] = df["paginas"].apply(
    lambda x: "Longo" if x > 350 else "Médio" if x > 150 or x < 350 else "Baixo")
```

```python
df['decada'] = (df['ano'] // 10 * 10)
```

```python
df_limpo = df[df["paginas"] > 0].copy()
```

```python
df_limpo['decada'] = (df_limpo['ano'] // 10 * 10)
```

```python
df_limpo["faixa_paginas"] = df_limpo["paginas"].apply(
    lambda x: "Longo" if x > 350 else "Médio" if x > 150 or x < 350 else "Baixo")
```
